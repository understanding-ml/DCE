```python
import torch
import pandas as pd

# 设置固定概率分布
probs = torch.tensor([0.3, 0.7])
seed = 42

# 设置全局种子
torch.manual_seed(seed)

# 使用默认全局 generator 的采样结果
samples_default = []
for _ in range(5):
    samples_default.append(torch.multinomial(probs, 1).item())
torch.rand(1)  # 会推进 global generator，但不影响你的自定义 generator

# 创建独立的 generator（种子相同）
g = torch.Generator().manual_seed(seed)

# 使用自定义 generator 的采样结果
samples_custom = []
for _ in range(5):
    samples_custom.append(torch.multinomial(probs, 1, generator=g).item())

# 显示结果
df = pd.DataFrame({
    "采样轮次": [f"第{i+1}次" for i in range(5)],
    "默认 generator（全局）": samples_default,
    "自定义 generator": samples_custom
})

df
```

```python
import numpy as np
import torch

# 设置全局种子
np.random.seed(42)
torch.manual_seed(42)

print("🌐 Global np:", np.random.randint(0, 100))
print("🌐 Global torch:", torch.randint(0, 100, (1,)))

# 创建新 RNG
rng_np = np.random.RandomState(42)
rng_torch = torch.Generator().manual_seed(42)

print("🔒 Local np:", rng_np.randint(0, 100))
print("🔒 Local torch:", torch.randint(0, 100, (1,), generator=rng_torch))
```

```python
np.random.seed(42)
torch.manual_seed(42)
rng_np = np.random.RandomState(42)
rng_torch = torch.Generator().manual_seed(42)

# 额外消耗一次全局 RNG 状态
np.random.randint(0, 100)
torch.randint(0, 100, (1,))

# 再采样
print("🌐 Global np:", np.random.randint(0, 100))
print("🌐 Global torch:", torch.randint(0, 100, (1,)))
print("🔒 Local np:", rng_np.randint(0, 100))
print("🔒 Local torch:", torch.randint(0, 100, (1,), generator=rng_torch))
```
